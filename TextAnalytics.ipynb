```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download(['punkt_tab', 'stopwords', 'wordnet', 'averaged_perceptron_tagger_eng'])
```

```python
# 1. Setup

text = "The students are studying and crying. My name is Obaidullah"

# 2. Tokenize & Clean Stop Words
tokenized_word=word_tokenize(text)
print(tokenized_word)
print(sent_tokenize(text))
```

```python
stop_words = set(stopwords.words('english'))
words = [w.lower() for w in tokenized_word if w.isalpha()]
print(words)
filtered = [w for w in words if w not in stop_words]
print(filtered)
```

```python
# 3. Stem & Lemma
ps = PorterStemmer()
words = ["wait", "waiting", "waited", "waits"]

print("\nStemming:")
for w in words:
    print(w, "->", ps.stem(w))
```

```python
lemmatizer = WordNetLemmatizer()
text3 = "studies studying cries cry"
tokens = word_tokenize(text3)

print("\nLemmatization:")
for w in tokens:
    print("Lemma for", w, "is", lemmatizer.lemmatize(w))
```

```python
print("POS Tags:", nltk.pos_tag(words))
```

```python
ps = PorterStemmer()
wnl = WordNetLemmatizer()

print("Stemmed:", [ps.stem(w) for w in filtered])
print("Lemma:", [wnl.lemmatize(w, pos='v') for w in filtered])
```
